```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings('ignore')

print("All Libraries Imported Successfully!")
```

```python
np.random.seed(42)

n = 1500

products = {
    "Laptop": ("Electronics", 50000, 38000),
    "Mobile": ("Electronics", 20000, 14000),
    "Mouse": ("Electronics", 500, 200),
    "Keyboard": ("Electronics", 1200, 600),
    "Monitor": ("Electronics", 12000, 8500),
    "Chair": ("Furniture", 2500, 1500),
    "Table": ("Furniture", 6000, 3500),
    "Sofa": ("Furniture", 25000, 18000),
    "Shirt": ("Clothing", 800, 400),
    "Jeans": ("Clothing", 1500, 800),
    "Shoes": ("Clothing", 2500, 1300),
    "TShirt": ("Clothing", 500, 250),
    "Headphones": ("Electronics", 3000, 1800),
    "Bookshelf": ("Furniture", 4500, 2800),
    "Jacket": ("Clothing", 3500, 1900)
}

product = np.random.choice(list(products.keys()), n)

category = [products[p][0] for p in product]
unit_price = np.array([products[p][1] for p in product])
cost = np.array([products[p][2] for p in product])

quantity = np.random.randint(1, 11, n)
discount = np.random.choice([0, 5, 10, 15, 20], n)

region = np.random.choice(["West", "North", "South", "East"], n)

city_map = {
    "West": ["Mumbai", "Pune", "Nashik"],
    "North": ["Delhi", "Jaipur", "Chandigarh"],
    "South": ["Bangalore", "Chennai", "Hyderabad"],
    "East": ["Kolkata", "Patna", "Bhubaneswar"]
}

city = [np.random.choice(city_map[r]) for r in region]

dates = pd.to_datetime(
    np.random.choice(
        pd.date_range("2025-01-01", "2025-12-31"),
        n
    )
)

sales = quantity * unit_price * (1 - discount / 100)
total_cost = quantity * cost
profit = sales - total_cost

df = pd.DataFrame({
    "Order_ID": range(10001, 10001+n),
    "Order_Date": dates,
    "Customer_ID": np.random.randint(1001, 1301, n),
    "Customer_Type": np.random.choice(["New", "Returning"], n),
    "Gender": np.random.choice(["Male", "Female"], n),
    "Region": region,
    "City": city,
    "Category": category,
    "Product": product,
    "Quantity": quantity,
    "Unit_Price": unit_price,
    "Discount_Percent": discount,
    "Cost_Per_Unit": cost,
    "Payment_Mode": np.random.choice(["UPI", "Cash", "Card", "Net Banking"], n),
    "Sales": sales,
    "Total_Cost": total_cost,
    "Profit": profit,
    "Order_Status": np.random.choice(
        ["Delivered", "Pending", "Cancelled"],
        n,
        p=[0.80, 0.12, 0.08]
    )
})

print("Dataset Created Successfully!")
```

```python
reference_date = df["Order_Date"].max() + pd.Timedelta(days=1)

rfm = df.groupby("Customer_ID").agg(
    Recency=("Order_Date", lambda x: (reference_date - x.max()).days),
    Frequency=("Order_ID", "nunique"),
    Monetary=("Sales", "sum")
)

rfm.head(10)
```

```python
rfm["R_Score"] = pd.qcut(
    rfm["Recency"].rank(method="first"),
    4,
    labels=[4, 3, 2, 1]
).astype(int)

rfm["F_Score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"),
    4,
    labels=[1, 2, 3, 4]
).astype(int)

rfm["M_Score"] = pd.qcut(
    rfm["Monetary"].rank(method="first"),
    4,
    labels=[1, 2, 3, 4]
).astype(int)

rfm.head()
```

```python
def customer_segment(row):

    if row["R_Score"] >= 3 and row["F_Score"] >= 3:
        return "Loyal Customers"

    elif row["R_Score"] >= 3 and row["M_Score"] >= 3:
        return "Potential Customers"

    elif row["R_Score"] <= 2:
        return "At Risk Customers"

    else:
        return "Regular Customers"

rfm["Segment"] = rfm.apply(customer_segment, axis=1)

rfm.head(10)
```

```python
rfm.groupby("Segment").agg(
    Total_Customers=("Monetary", "count"),
    Total_Revenue=("Monetary", "sum"),
    Average_Revenue=("Monetary", "mean")
)
```

```python
plt.figure(figsize=(10, 6))

sns.countplot(
    data=rfm,
    x="Segment",
    order=rfm["Segment"].value_counts().index
)

plt.title("Customer Segmentation Analysis")
plt.xticks(rotation=25)
plt.ylabel("Number of Customers")
plt.show()
```

```python
Q1 = df["Sales"].quantile(0.25)
Q3 = df["Sales"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[
    (df["Sales"] < lower_limit) |
    (df["Sales"] > upper_limit)
]

print("Total Outliers:", len(outliers))
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
```
